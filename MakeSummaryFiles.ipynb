```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import glob
```

```python
regen_files = True
```

## 36 Oph Parameters and Summary Plots

```python
## change this folder to the sim set the summary file should be generated for:
filedir = 'results/36OphA_45deg/'
filelist = glob.glob(filedir + "*csv")
if regen_files == True:
    masterdata_list = pd.DataFrame()
    for item in filelist:
        data = pd.read_csv(item)
        binary_orbit = data.loc[(data['hash'] == "36OphB") & (data['primary']==0)]
        data['binary_sma'] = binary_orbit['a'].values[0]
        data['binary_ecc'] = binary_orbit['e'].values[0]
        data['sourcefile'] = item
        data_subset = data.loc[data['time'] == 0] # take initial time step only to make file size manageable for summary files
        masterdata_list = pd.concat([masterdata_list,data_subset], ignore_index=True)
```

```python
import matplotlib.pyplot as plt

# Take the first value for each unique sourcefile
# (binary orbit properties are the same within a single file)
sma_per_file = (masterdata_list.groupby('sourcefile')['binary_sma'].first())
ecc_per_file = (masterdata_list.groupby('sourcefile')['binary_ecc'].first())

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
fig.suptitle("36 Oph, Binary Orbit Across All Simulation Sets")
ax[0].hist(sma_per_file.dropna(), bins=15)
ax[0].set_xlabel('Binary Orbit Semi-Major Axis')
ax[0].set_ylabel('N')
ax[1].hist(ecc_per_file.dropna(), bins=15)
ax[1].set_xlabel('Binary Orbit Eccentricity')
ax[1].set_ylabel('N')

plt.tight_layout()
```

```python
# Keep only planet particles
planets = masterdata_list[masterdata_list['hash'].astype(str).str.contains('test', na=False)]

fig, ax = plt.subplots(1, 3, figsize=(14, 4))

fig.suptitle("36 Oph, Planet Particle Initial Orbits Across All Simulation Sets")

# SMA
ax[0].hist(planets['a'].dropna(), bins=15)
ax[0].set_xlabel('Planet Initial Semi-Major Axis [AU]')
ax[0].set_ylabel('N')

# ecc
ax[1].hist(planets['e'].dropna(), bins=15)
ax[1].set_xlabel('Planet Initial Eccentricity')
ax[1].set_ylabel('N')

# Inclination
ax[2].hist(planets['inc'].dropna() * 180/ np.pi, bins=15)
ax[2].set_xlabel('Planet Initial Inclination (deg)')
ax[2].set_ylabel('N')

plt.tight_layout()
```
